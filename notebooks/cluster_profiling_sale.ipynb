```python
# In your Jupyter Notebook

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
from pathlib import Path
from IPython.display import display

# --- Configuration ---
# Define paths assuming this notebook is in the 'notebooks/' directory
NOTEBOOK_DIR = Path.cwd()
PROJECT_ROOT = NOTEBOOK_DIR.parent
PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"

# Load the CSV file that contains the final cluster labels
CLUSTERED_FILENAME = "sale_properties_clustered.csv"
CLUSTERED_FILEPATH = PROCESSED_DATA_DIR / CLUSTERED_FILENAME

# --- Load the Data ---
try:
    df_clustered = pd.read_csv(CLUSTERED_FILEPATH)
    print("Clustered data loaded successfully.")
    print("Shape:", df_clustered.shape)
    display(df_clustered.head())
except FileNotFoundError:
    print(f"ERROR: Clustered data file not found at {CLUSTERED_FILEPATH}")
```

```python
print("--- Cluster Size Distribution ---")
cluster_counts = df_clustered['cluster'].value_counts().sort_index()
print(cluster_counts)

# Visualize the distribution
fig = px.bar(
    cluster_counts, 
    x=cluster_counts.index, 
    y=cluster_counts.values,
    title="Number of Properties per Cluster",
    labels={'x': 'Cluster ID', 'y': 'Number of Properties'}
)
fig.show()
```

```python
print("--- Analysis of Numerical Features by Cluster ---")

# List of key numerical features to analyze
numerical_features = [
    'price_eur',
    'superficie_construida',
    'habitaciones',
    'banos',
    'planta_numerica',
    'antiguedad', # This is now numerical (ordinal)
    'conservacion', # This is now numerical (ordinal)
    'barrio_encoded', # The target-encoded value
]

# Group by cluster and calculate the mean and median for each feature
# Using .agg() allows us to calculate multiple statistics at once
cluster_summary_numerical = df_clustered.groupby('cluster')[numerical_features].agg(['mean', 'median', 'count'])

# Display the summary table
# We'll round the values for easier reading
display(cluster_summary_numerical.round(2))
```

```python
print("\n--- Analysis of Boolean (Amenity) Features by Cluster ---")

# List of key boolean features
boolean_features = [
    'ascensor',
    'piscina',
    'garaje',
    'terraza',
    'trastero',
    'exterior',
    # Include the one-hot encoded columns for 'amueblado'
    'amueblado_True',
    'amueblado_False',
]

# Group by cluster and calculate the mean (percentage) for each boolean feature
cluster_summary_boolean = df_clustered.groupby('cluster')[boolean_features].mean()

# Display the summary table (values will be from 0.0 to 1.0)
display((cluster_summary_boolean * 100).round(2).style.background_gradient(cmap='viridis').format("{:.2f}%"))
```

## Proposed Labels and Profiles for Each Cluster


### Cluster 1: Standard & Affordable Apartments
This is the largest group (1365 properties), representing the most common type of property on the market.

#### Defining Characteristics:

*   **Price & Value**: Has the lowest average price (~€626k) and is located in neighborhoods with the lowest average property value (barrio_encoded).
*   **Size**: Has the smallest average size (~96 m²) and fewest rooms/bathrooms.
*   **Amenities**: Lacks most luxury amenities. It has the lowest percentage of elevators (33%), pools (1.6%), and garages (11%).
*   **Condition**: Generally older properties in "good condition" but typically not recently renovated.
*   **Persona**: This cluster represents the core of the Madrid housing market: standard, smaller apartments in less central or less expensive neighborhoods. They are functional homes but generally lack modern luxuries like pools, garages, or guaranteed elevators, which is typical for a large portion of the city's building stock.


### Cluster 0: Classic, Large City Center Apartments
This is a significant group of 660 properties that represents a very classic Madrid property type.

#### Defining Characteristics:

*   **Price & Value**: High price (~€1.6M) and located in high-value neighborhoods.
*   **Size**: Quite large (~167 m²) with 3+ rooms and 2.5+ bathrooms on average.
*   **Amenities**: Very high chance of having an elevator (93%) and being an exterior-facing unit (75%). However, it has a very low percentage of pools and garages.
*   **Condition**: Tends to be in older buildings (antiguedad mean is very low).
*   **Persona**: This cluster perfectly describes the classic, spacious apartments found in the established, central districts of Madrid (often called a "piso señorial"). These are desirable properties in older, well-maintained buildings that have elevators but typically lack the space for amenities like private pools or large garages, which are rare in historic city centers.


### Cluster 4: Premium Renovated & Furnished Apartments
This group of 587 properties represents a high-end, "turn-key" segment of the market.

#### Defining Characteristics:

*   **Price & Value**: High price (~€1.5M), located in the most expensive neighborhoods (highest barrio_encoded value of the main groups).
*   **Condition**: These are, on average, the oldest properties but have the best conservation status, strongly indicating they have been recently renovated.
*   **Amenities**: High chance of an elevator (78%). A defining feature is that 100% of these properties are furnished (amueblado_True).
*   **Size**: Medium-sized (~133 m²), smaller than the "classic" apartments in Cluster 0.
*   **Persona**: This cluster represents apartments in prime, historic neighborhoods that have been fully renovated to a high standard and are sold ready to be lived in. They are perfect for buyers looking for a premium, hassle-free property in a top location, or for investors seeking a high-yield rental property.

### Cluster 3: Modern Luxury Homes & Penthouses
This group of 388 properties clearly represents the modern, high-end luxury market.

#### Defining Characteristics:

*   **Price & Value**: The highest average price of all major clusters (~€1.9M).
*   **Size**: Very large (~257 m²) with the most rooms and bathrooms (~4.25 and ~3.7 respectively).
*   **Amenities**: This cluster is defined by its luxury amenities. It has by far the highest percentage of pools (81%) and garages (84%), along with a high likelihood of having a terrace and storage room (trastero).
*   **Condition**: These are the newest properties on average (antiguedad score is highest).
*   **Persona**: This cluster represents modern luxury. These are likely penthouses, duplexes, or houses/chalets in newer, exclusive residential complexes (urbanizaciones) where amenities like pools, private garages, and large terraces are standard features.
