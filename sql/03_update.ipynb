#### Insertar o actualizar recorriendo el df

```python
# funcion auxiliar que inserta valores si no existen 
# y devuelve el id creado o ya existente
def get_or_create(table, keycol, keyval, idcol, extras=None):
    # intenta SELECT
    sql = f"SELECT {idcol} FROM {table} WHERE {keycol}=%s"
    cursor.execute(sql, (keyval,))
    r = cursor.fetchone()
    if r:
        return r[0]
    # si no existe, INSERT
    cols = [keycol] + (list(extras.keys()) if extras else [])
    vals = [keyval] + (list(extras.values()) if extras else [])
    ph = ",".join(["%s"]*len(vals))
    sql = f"INSERT INTO {table} ({','.join(cols)}) VALUES ({ph})"
    cursor.execute(sql, vals)
    return cursor.lastrowid
```

```python
bool_feats = [
    'ascensor','balcon','calefaccion','chimenea','cocina_equipada','exterior',
    'garaje','jardin','piscina','trastero','terraza','adaptado_movilidad_reducida',
    'aire_acondicionado','puerta_blindada','vidrios_dobles','sistema_seguridad'
]
```

```python
df_all = df_all.replace({np.nan:None})
```

```python
for idx, row in df_all.iterrows():
    url = row['url']
    
    # 7.1 LOCATION
    loc_id = get_or_create(
        'location','barrio', row['barrio'], 'location_id',
        extras={'distrito': row['distrito']}
    )
    
    # 7.2 AGE_RANGE
    age_id = None
    if pd.notna(row['antiguedad']):
        age_id = get_or_create('age_range','label', row['antiguedad'], 'age_range_id')
    
    # 7.3 PROPERTY (upsert via ON DUPLICATE KEY)
    pk_native = row['property_id']
    prop_cols = [
        'location_id','latitude','longitude','property_native_id',
        'superficie_construida','superficie_util','habitaciones','banos',
        'planta','estado_conservacion','age_range_id'
    ]
    prop_vals = [
        loc_id, row['latitude'], row['longitude'], pk_native,
        row['superficie_construida'], row['superficie_util'],
        row['habitaciones'], row['banos'], row['planta_numerica'],
        row['conservacion'], age_id
    ]
    upd = ",".join(f"{c}=VALUES({c})" for c in prop_cols[1:])
    sql_prop = (
        f"INSERT INTO property ({','.join(prop_cols)}) "
        f"VALUES ({','.join(['%s']*len(prop_vals))}) "
        f"ON DUPLICATE KEY UPDATE {upd}"
    )
    cursor.execute(sql_prop, prop_vals)
    # recuperar el property_id real
    if pk_native in map_prop:
        property_id = map_prop[pk_native]
    else:
        property_id = cursor.lastrowid
        map_prop[pk_native] = property_id

    # 7.4 ENERGY_CERTIFICATE
    if pd.notna(row['energy_consumption_kwh_m2_yr']) or pd.notna(row['energy_emissions_kg_co2_m2_yr']):
        sql_ec = """
        INSERT INTO energy_certificate
          (property_id,classification,consumo_rating,emisiones_rating,emision_value,consumption_value)
        VALUES (%s,%s,%s,%s,%s,%s)
        ON DUPLICATE KEY UPDATE
          classification=VALUES(classification),
          consumo_rating=VALUES(consumo_rating),
          emisiones_rating=VALUES(emisiones_rating),
          emision_value=VALUES(emision_value),
          consumption_value=VALUES(consumption_value)
        """
        cursor.execute(sql_ec, (
            property_id,
            row['energy_cert_classification'],
            row['energy_consumption_rating'],
            row['energy_emissions_rating'],
            row['energy_emissions_kg_co2_m2_yr'],
            row['energy_consumption_kwh_m2_yr'],
        ))

    # 7.5 FEATURES EAV
    for feat in bool_feats:
        if row.get(feat) is True:
            # obtener o crear feature_id
            if feat not in feat_map:
                fid = get_or_create('feature_catalog','nombre',feat,'feature_id')
                feat_map[feat] = fid
            else:
                fid = feat_map[feat]
            # insertar en property_feature
            cursor.execute(
                "INSERT IGNORE INTO property_feature(property_id,feature_id) VALUES(%s,%s)",
                (property_id, fid)
            )

    # 7.6 LISTING INSERT / UPDATE
    price = row['price_eur']
    scraped_at = row['scraped_at']
    desc = row['description']
    if url in map_list:
        # UPDATE
        lid = map_list[url]
        cursor.execute("""
            UPDATE listing
               SET price_eur=%s, scraped_at=%s, scrape_status=%s, description=%s
             WHERE listing_id=%s
        """, (price, scraped_at, 'Success', desc, lid))
    else:
        # INSERT
        sql_l = """
        INSERT INTO listing
          (property_id,url,listing_type,price_kind,price_eur,scraped_at,scrape_status,description)
        VALUES (%s,%s,%s,%s,%s,%s,%s,%s)
        """
        cursor.execute(sql_l, (
            property_id, url, row['listing_type'], row['price_kind'],
            price, scraped_at, 'Success', desc
        ))
        map_list[url] = cursor.lastrowid

    # Commit cada 200 filas
    if idx % 200 == 0:
        conn.commit()

# commit final
conn.commit()
print("Insert/Update completado.")
```

```python
if to_remove:
    params = tuple(to_remove)
    sql = f"UPDATE listing SET scrape_status='Removed' WHERE url IN ({','.join(['%s']*len(params))})"
    cursor.execute(sql, params)
    conn.commit()
    print(f"{len(params)} anuncios marcados como Removed.")
else:
    print("No hay anuncios para marcar como Removed.")
```

```python
cursor.close()
conn.close()
```

Creamos tres df, los anuncios nuevos (a insertar), los anuncios existentes (a actualizar), y los que ya no existen (marcar como borrados-remove)

```python
urls_nuevas   = set(df_all['url']) - set(map_list.keys())
urls_existentes = set(df_all['url']) & set(map_list.keys())
urls_remover = set(map_list.keys()) - set(df_all['url'])

print(f"{len(urls_nuevas)} nuevas, {len(urls_existentes)} existentes, {len(urls_remover)} a marcar removed")
```

```python
df_insert = df_all[df_all['url'].isin(urls_nuevas)].copy()
df_insert
```

```python
df_update = df_all[df_all['url'].isin(urls_existentes)].copy()
df_update
```

```python
df_remove = (pd.DataFrame({'url': list(urls_remover)})
               .assign(listing_id=lambda d: d['url'].map(map_list))
            )
df_remove
```

Marcamos los anuncios borrados como "removed"

```python
if not df_remove.empty:
    ids = df_remove['listing_id'].tolist()
    sql = f"UPDATE listing SET scrape_status='Removed' WHERE listing_id IN ({','.join(['%s']*len(ids))})"
    cursor.execute(sql, ids)
    conn.commit()
    print(f"Marcados {len(ids)} listings como Removed")
```
